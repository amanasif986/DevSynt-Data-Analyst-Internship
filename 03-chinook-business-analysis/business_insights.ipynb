# Business Insights Chinook Store Analysis



## 1. The USA is the anchor market

### Finding

The USA is the top market, bringing in more revenue and more customers than any other country.

### Supporting Value

USA revenue is $523.06 from 13 customers.

### Recommended Action

Focus marketing and retention efforts on the USA market.


## 2. Rock accounts for over a third of all revenue

### Finding

Rock is the best-selling genre by a wide margin.

### Supporting Value

Rock generated $826.65 in revenue, far ahead of every other genre.

### Recommended Action

Keep a strong Rock catalog while working to grow genres like Latin and Metal to reduce reliance on one category.


## 3. Revenue is essentially flat across five years, with a recent dip

### Finding

Revenue held steady for four years, then slipped in the most recent year.

### Supporting Value

Revenue stayed around $450 to $480 per year from 2021 through 2024, then dropped to $450.58 in 2025.

### Recommended Action

Investigate what changed in 2025 so the dip doesn't continue.


## 4. Purchase frequency is almost perfectly uniform

### Finding

Almost every customer buys the same number of times.

### Supporting Value

Nearly all 59 customers made exactly 7 purchases.

### Recommended Action

Focus on increasing spend per order rather than purchase frequency, since frequency is already high and consistent.


## 5. Revenue is broadly distributed, not concentrated in a small VIP group

### Finding

No small group of customers dominates sales; spending is spread out fairly evenly.

### Supporting Value

The top 10 customers account for just 19.4 percent of total revenue.

### Recommended Action

Run a store-wide loyalty program rather than a VIP-only program.
