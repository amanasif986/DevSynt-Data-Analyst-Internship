## 2. Identifying data quality issues.

```python
# Check missing values
missing = df.isnull().sum()
missing_summary = pd.DataFrame({'Missing Count': missing})
missing_summary[missing_summary['Missing Count'] > 0]
```

```python
# Check for duplicate rows
duplicate = df.duplicated().sum()
print(f"There are {duplicate} duplicate rows in the dataset.")
```

```python
# Check unique values in categorical columns
print("Sex:", df['Sex'].unique())
print("Embarked:", df['Embarked'].unique())
print("Pclass:", df['Pclass'].unique())
print("Survived:", df['Survived'].unique())
```

## 3. Handling missing values using appropriate techniques.

```python
# Fill missing values in 'Age' column with median because it is a continuous variable and median is less affected by outliers than mean.
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
# Fill missing values in 'Embarked' column with mode because it is a categorical variable and mode is the most frequent value.
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
# Fill missing values in 'Cabin' column with 'Unknown' because it is a categorical variable and we want to keep track of missing values.
df['Cabin'] = df['Cabin'].fillna('Unknown')
```

```python
# Check for missing values after filling
df.isnull().sum()
```

## 4.Detecting and removing duplicate records

```python
print(f"Duplicate rows found: {duplicate}")

if duplicate > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")
else:
    print("No duplicate rows to remove.")
```

## 5. Changing data types of columns to appropriate types.

```python
df['Survived'] = df['Survived'].astype('category')
df['Pclass'] = df['Pclass'].astype('category')
df['Sex'] = df['Sex'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')

df.dtypes
```

## 6. Renaming columns where necessary to improve readability.

```python
df = df.rename(columns={
    'SibSp': 'Sibling or Spouses Aboard',
    'Parch': 'Parents or Children Aboard',
    'Pclass': 'Passenger Class',
    'Embarked': 'Embarkation'})
df.columns
```

## 7. Making it more readable


```python
df["Sex"] = df["Sex"].map({"male":"Male", "female":"Female"})
df["Embarkation"] = df["Embarkation"].map({"C":"Cherbourg", "Q":"Queenstown", "S":"Southampton"})
df["Passenger Class"] = df["Passenger Class"].map({1:"First Class", 2:"Second Class", 3:"Third Class"})
df["Survived"] = df["Survived"].map({0:"No", 1:"Yes"})
df.head()
```

```python
df['Fare'] = df['Fare'].round(2)
df.head()
```

## 8. Saving the cleaned dataset to a new CSV file for future use.

```python
df.to_csv('cleaned_titanic.csv', index=False)
```
